# P4 · tgbn-token: распределение m_msg-скоров при log1p-трансформе

**Зачем.** На token метод (`--m_message_feature`) не даёт прироста. Прошлые замеры показали, что скор-колонка там **бимодальна и zero-inflated** (Сарл 0.59–0.69, 15.5% near-zero на предсказываемых юзерах), а два канала ведут себя противоположно. Здесь смотрим на это внимательно и именно в том виде, в каком оно идёт в модель — **после log1p**.

**Что именно измеряем** (κ=16 — оптимум token):
- **memory-канал** = `log1p(score_for)` на записываемых рёбрах (то, что уходит в GRU-память);
- **GNN-канал** = `log1p(m_e_raw)` на top-k выбранных item'ах **предсказываемых** (label) юзеров (то, что реально влияет на предсказание);
- форма (гистограммы, Сарл, доля near-zero), дрейф train→val→test, и **сколько label-юзеров имеют почти пустую строку `M`** (структурная причина zero-inflation — транзитные юзеры token, медиана 158 событий).

```
# Ячейка 1 — стрим token, сбор сырых скоров обоих каналов + пустых строк M по сплитам (κ=16)
import numpy as np, polars as pl, torch, math, time
import plotly.express as px, plotly.graph_objects as go
from scipy import stats
from tgb.nodeproppred.dataset_pyg import PyGNodePropPredDataset
from torch_geometric.loader import TemporalDataLoader
import sys; sys.path.insert(0, "..") if ".." not in sys.path else None
from models.msampler import MSampler

ds = PyGNodePropPredDataset(name="tgbn-token", root="datasets"); data = ds.get_TemporalData()
tr, va, te = data.train_val_test_split(val_ratio=0.15, test_ratio=0.15)
N, C = data.num_nodes, ds.num_classes
smp = MSampler(N, C, 10, math.log(2)/(16*86400.0), torch.sort(tr.msg[:, 0]).values, device="cpu")
n_tr = tr.src.size(0); cut = int(n_tr*0.85)   # late-train порог для сэмплинга
mem = {"train": [], "val": [], "test": []}     # score_for на записываемых рёбрах
gnn = {"train": [], "val": [], "test": []}     # top-k decayed M на label-юзерах
emptyrow = {"train": [], "val": [], "test": []}# доля label-юзеров с пустой строкой M
support = {"train": [], "val": [], "test": []} # размер support строки (ненулевых item)

def read_rows(us):  # реплика __call__: decayed M[us] -> топ-k>0 + support
    r = smp.M[us].clone()
    if smp.alpha > 0:
        r = r * torch.exp(-smp.alpha*(smp.t_now - smp.tlast[us]).clamp(min=0))
    supp = (r > 0).sum(1)
    topv = r.topk(min(10, C), dim=1).values
    return topv[topv > 0], supp

t0 = time.time(); seen_edges = 0; day = 0
def run(split, dd, sample_every, bs=5000):
    global day, seen_edges
    label_t = ds.get_label_time()
    for b in TemporalDataLoader(dd, batch_size=bs):
        seen_edges += b.src.numel()
        if float(b.t[-1]) > label_t:
            lt = ds.get_node_label(b.t[-1])
            if lt is None: break
            l0 = float(lt[0][0]); us = lt[1]; label_t = ds.get_label_time(); day += 1
            pm = b.t < l0
            collect = (split != "train" and day % sample_every == 0) or (split == "train" and seen_edges >= cut and day % sample_every == 0)
            if pm.any():
                if collect: mem[split].append(smp.score_for(b.src[pm], b.dst[pm], b.t[pm]).numpy())
                smp.insert(b.src[pm], b.dst[pm], b.t[pm], b.msg[pm])
            if collect:
                gv, supp = read_rows(us)
                gnn[split].append(gv.numpy()); support[split].append(supp.numpy())
                emptyrow[split].append(float((supp == 0).float().mean()))
            m2 = ~pm
            if m2.any(): smp.insert(b.src[m2], b.dst[m2], b.t[m2], b.msg[m2])
        else:
            smp.insert(b.src, b.dst, b.t, b.msg)
ds.reset_label_time()
run("train", tr, 8); run("val", va, 3); run("test", te, 3); ds.reset_label_time()
for k in mem: mem[k] = np.concatenate(mem[k]) if mem[k] else np.array([]); gnn[k] = np.concatenate(gnn[k]) if gnn[k] else np.array([]); support[k] = np.concatenate(support[k]) if support[k] else np.array([])
print(f"[{time.time()-t0:.0f} c] собрано mem: { {k: len(v) for k,v in mem.items()} }")
print(f"           собрано gnn: { {k: len(v) for k,v in gnn.items()} }")
print(f"пустых строк M у label-юзеров: " + "  ".join(f"{k}={np.mean(emptyrow[k]):.1%}" for k in emptyrow))
```

```
# Ячейка 2 — форма log1p-скоров обоих каналов (val = eval-плато) + Сарл + near-zero
def sarle(x):
    n=len(x); g=stats.skew(x); k=stats.kurtosis(x)
    return (g**2+1)/(k+3*(n-1)**2/((n-2)*(n-3)))
from plotly.subplots import make_subplots
fig = make_subplots(rows=1, cols=2, subplot_titles=["memory-канал (записываемые рёбра)", "GNN-канал (top-k → предсказание)"])
for c,(ch,arr) in enumerate([("memory", mem["val"]), ("GNN", gnn["val"])], start=1):
    l = np.log1p(arr)
    fig.add_histogram(x=l, nbinsx=70, histnorm="probability density", marker_color=["#2c7fb8","#d7191c"][c-1], row=1, col=c, showlegend=False)
    q = np.percentile(l, [5,50,95])
    print(f"{ch:7} log1p: p5={q[0]:.2f} p50={q[1]:.2f} p95={q[2]:.2f} | доля(log1p≤0.1)={ (l<=0.1).mean():.1%} "
          f"доля(log1p≤0.5)={(l<=0.5).mean():.1%} | Сарл={sarle(l):.3f}" + ("  ←БИМОДАЛЬНО" if sarle(l)>0.555 else ""))
fig.update_xaxes(title_text="log1p(score)"); fig.update_layout(height=420, title="token: log1p-скоры — memory ГОРЯЧИЙ и правый, GNN холодный со спайком у нуля")
fig.show()
# для контекста: масштаб остальных фич модели
print("справка: raw_msg (вес ребра) p95 =", float(np.percentile(tr.msg[:,0].numpy(),95)), "; cos-энкодинги ∈ [-1,1]")
```

```
# Ячейка 3 — дрейф log1p-скоров train→val→test (нестационарность поверх бимодальности)
rows = []
for ch, dd in [("memory", mem), ("GNN", gnn)]:
    for sp in ["train", "val", "test"]:
        l = np.log1p(dd[sp])
        rows.append((ch, sp, np.median(l), np.percentile(l, 95), (l <= 0.5).mean()))
dr = pl.DataFrame(rows, schema=["канал", "сплит", "log_p50", "log_p95", "доля≤0.5"], orient="row")
print(dr)
fig = go.Figure()
for ch, col in [("memory", "#2c7fb8"), ("GNN", "#d7191c")]:
    sub = dr.filter(pl.col("канал") == ch)
    fig.add_scatter(x=sub["сплит"], y=sub["log_p50"], mode="lines+markers", name=f"{ch} p50", line=dict(color=col))
    fig.add_scatter(x=sub["сплит"], y=sub["log_p95"], mode="lines+markers", name=f"{ch} p95", line=dict(color=col, dash="dot"))
fig.update_layout(title="token: медиана/p95 log1p-скора по сплитам — memory дрейфует вверх, GNN сидит внизу",
                  yaxis_title="log1p(score)", height=400)
fig.show()
print(f"\nразрыв каналов по медиане (val): memory {np.median(np.log1p(mem['val'])):.2f} vs GNN {np.median(np.log1p(gnn['val'])):.2f}"
      f"  → ×{np.median(np.log1p(mem['val']))/max(np.median(np.log1p(gnn['val'])),1e-9):.0f} в log-пространстве")
```
